# L4 - Validation de la qualité des données avec Great Expectations

## Objectif

L’objectif de ce livrable est de garantir la qualité et la fiabilité des données analytiques générées dans la couche Gold à l’aide de Great Expectations.

Cette validation permet de vérifier :
- la cohérence des données,
- l’absence de valeurs invalides,
- la qualité des indicateurs analytiques,
- ainsi que la fiabilité des datasets utilisés pour les dashboards et analyses métier.


```python
# Importation des bibliothèques

import pandas as pd
import great_expectations as gx
```

```python
# Initialisation du contexte Great Expectations

context = gx.get_context(mode="file")

print("Contexte Great Expectations initialisé")
```

```python
# Chargement du dataset Gold

df = pd.read_csv("../data/gold/rfm_customers.csv")
df.columns = df.columns.str.lower()

df.head()
```

```python
# Informations sur le dataset

df.info()
```

```python
print("\nColonnes après normalisation :")
print(df.columns)
```

```python
# Validation de CustomerID
null_customer_ids = df["customerid"].isnull().sum()

print("Valeurs nulles customerid :", null_customer_ids)
```

```python
# Validation des valeurs Monetary

monetary_check = (df["monetary"] >= 0).all()

print("Monetary positif :", monetary_check)
```

```python
# Validation des valeurs Frequency

frequency_check = (df["frequency"] >= 1).all()

print("Frequency valide :", frequency_check)
```

```python
# Validation des valeurs Recency

recency_check = (df["recency"] >= 0).all()

print("Recency valide :", recency_check)
```

```python
# Validation des segments clients

valid_segments = [
    "Champions",
    "Loyal Customers",
    "Potential Loyalists",
    "New Customers",
    "At Risk",
    "Hibernating",
    "Lost",
    "Others"
]

if "segment" in df.columns:
    
    # Afficher les segments présents dans le dataset
    print("Segments présents dans le dataset :")
    print(df["segment"].unique())

    # Vérification
    segment_check = df["segment"].isin(valid_segments).all()

    print("\nSegments valides :", segment_check)

else:
    
    print("La colonne segment n'existe pas encore dans le dataset")
```

```python
# Résumé global des validations
print("\n----- Résumé des validations -----\n")

print("customer_id valide :", null_customer_ids == 0)

print("Monetary valide :", monetary_check)

print("Frequency valide :", frequency_check)

print("Recency valide :", recency_check)

print("Segments valides :", segment_check)
```
